```python
faqs = [
    {
        "question": "What are the college admission requirements?",
        "answer": "You need to meet the eligibility criteria and submit the required documents."
    },
    {
        "question": "How can I apply for admission?",
        "answer": "You can apply for admission through the college's official admission portal."
    },
    {
        "question": "What courses does the college offer?",
        "answer": "The college offers various undergraduate and postgraduate courses."
    },
    {
        "question": "What is the college fee?",
        "answer": "The fee depends on the course and program you choose."
    },
    {
        "question": "Does the college provide hostel facilities?",
        "answer": "Yes, hostel facilities are available for eligible students."
    },
    {
        "question": "How can I contact the college?",
        "answer": "You can contact the college through its official website or office phone number."
    }
]

print(faqs)
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('stopwords')

print("NLP libraries loaded successfully!")
```

```python
nltk.download('punkt_tab')
```

```python
import string

def clean_text(text):
    text = text.lower()
    text = text.translate(str.maketrans('', '', string.punctuation))
    words = word_tokenize(text)
    words = [word for word in words if word not in stopwords.words('english')]
    return " ".join(words)

print(clean_text("What are the college admission requirements?"))
```

```python
faq_questions = [clean_text(faq["question"]) for faq in faqs]

print(faq_questions)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

vectorizer = TfidfVectorizer()

faq_vectors = vectorizer.fit_transform(faq_questions)

user_question = "How do I get admission?"
user_question_clean = clean_text(user_question)

user_vector = vectorizer.transform([user_question_clean])

similarity = cosine_similarity(user_vector, faq_vectors)

best_match = similarity.argmax()

print("User Question:", user_question)
print("Best Matching FAQ:", faqs[best_match]["question"])
print("Answer:", faqs[best_match]["answer"])
```

```python
def get_answer(user_question):
    cleaned_user_question = clean_text(user_question)

    user_vector = vectorizer.transform([cleaned_user_question])

    similarity = cosine_similarity(user_vector, faq_vectors)

    best_match = similarity.argmax()
    best_score = similarity[0][best_match]

    if best_score < 0.25:
        return "Sorry, I don't know the answer to that question."

    return faqs[best_match]["answer"]


print(get_answer("What is the college fee?"))
```
